```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

print("Modules imported successfully!")
```

```python
#1. Declare a basic Python List
depth_list = [10, 20, 30, 40, 50]

#2. Declare a NumPy array
temperature_array = np.array([15.2, 14.1, 11.5, 8.3, 4.2])

print("List:", depth_list)
print("NumPy Array:", temperature_array)
```

```python
def calculate_gradient(change_in_y, change_in_x):
    """A basic function to calculate a rate of change."""
    return change_in_y / change_in_x

# Test the function
rate = calculate_gradient((15.2 - 4.2), (50 - 10))
print(f"Calculated geothermal gradient: {rate:.2f}°C/m")
```

```python
# Create the plot figure
plt.figure(figsize=(6, 4))
plt.plot(depth_list, temperature_array, marker="o", color="darkred", linestyle="--")

# Label the axes
plt.title("Earth Science Profile: Temperature vs. Depth")
plt.xlabel("Depth (meters)")
plt.ylabel("Temperature (°C)")
plt.grid(True)

# Save the figure as a PNG file as required by the rubric
plt.savefig("mstrapason_plot.png", dpi=300)
plt.show()
```
